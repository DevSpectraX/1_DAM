1. Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def contarLetras(texto):
    frecuencias = {}

    for letra in texto:
        if letra != " ":
            if letra in frecuencias:
                frecuencias[letra] += 1
            else:
                frecuencias[letra] = 1
    
    return frecuencias

print(contarLetras("hola que tal"))
```

```python
numeros = [12, 7, 25, 12, 9, 18, 25, 4, 7, 31, 18, 42, 9, 50, 12]

def duplicados(lista):

    #Creamos un set
    nuevo_set = set()

    #Recorremos la lista
    for x in lista:
        #Si el elemento esta en el set, salimos del bucle y mostramos ese elemento
        if x in nuevo_set:
            return x
        #Si no está, lo añadimos
        nuevo_set.add(x)


print(duplicados(numeros))
```

30. Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
def anagrama(primera, segunda):

    #Darle la vuelta a la primera palabra
    palabra_reves = primera[::-1]

    #Comparar si la palabra al reves es igual a la segunda
    if palabra_reves == segunda:
        return "Es un anagrama"
    else: 
        return "No es un anagrama"

print(anagrama("amor", "roma"))
```

     - Código a seguir:
          - Crear una función contar_palabras que cuente el número de veces que aparece cada palabra en el texto y devuelva un diccionario.
          - Crear una función reemplazar_palabras para sustituir una palabra_original por una palabra_nueva en el texto y devolver el texto modificado.
          - Crear una función eliminar_palabra que elimine una palabra del texto y devuelva el texto sin ella.
          - Crear la función procesar_texto que reciba un texto, una opción ("contar", "reemplazar", "eliminar") y un número variable de argumentos según la opción elegida.
     - Caso de uso:
          - Verificar el funcionamiento completo de procesar_texto.


```python
def procesar_texto(texto, opcion, *palabras):

    def contador_palabras(texto):
        palabras = texto.split()
        contador = {}

        for palabra in palabras:
            if palabra in contador:
                contador[palabra] += 1
            else:
                contador[palabra] = 1
        return contador

    def reemplazar_palabras(texto, *palabras):
        palabra_original = palabras[0]
        palabra_nueva = palabras[1]

        texto_nuevo  = texto.replace(palabra_original, palabra_nueva)

        return texto_nuevo

    def eliminar_palabras(texto, *palabras):
        texto_nuevo = texto.replace(palabras[0], "")
        return texto_nuevo



    if opcion == "contar":
        return contador_palabras(texto)
    if opcion == "reemplazar":
        return reemplazar_palabras(texto, *palabras)
    if opcion == "eliminar":
        return eliminar_palabras(texto, *palabras)




    
        
frase = "Hoy fui al mercado porque necesitaba comprar pan. En el mercado compré pan, fruta y leche, y después volví a casa para preparar la comida."
procesar_texto(frase, "contar")
procesar_texto(frase, "reemplazar", "Hoy", "Mañana")
procesar_texto(frase, "eliminar", "mercado")


    
```

    - Solicitar al usuario el precio original de un artículo.
    - Preguntar si tiene un cupón de descuento (respuesta sí o no).
    - Si la respuesta es sí, solicitar el valor del cupón de descuento.
    - Aplicar el descuento al precio original, siempre que el valor del cupón sea válido (mayor a cero).
    - Mostrar el precio final de la compra, considerando o no el descuento.
    - Usar estructuras de control de flujo (if, elif, else) para llevar a cabo las acciones.

```python
precio_original = int(input("¿Cuál es el valor del producto?"))
tiene_cupon = input("¿Tienes un cupón de descuento?")

if tiene_cupon.lower() == "si":
    valor_cupon = int(input("¿De qué valor es el cupón?"))
    if valor_cupon > 0:
        precio_final = precio_original * (1 - valor_cupon / 100)
        print("El precio total aplicando tu cupón de", valor_cupon,"% es de:", precio_final)
    else:
        print("El precio del cupón es erroneo")    
elif tiene_cupon.lower() == "no":
    print("El precio total es: ", precio_original)
```
